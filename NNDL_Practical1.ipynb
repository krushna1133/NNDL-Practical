```python
import numpy as np
```

```python
#Vector addition
vector1 = np.array([10,20,30])
vector2 = np.array([5,10,15])
print("Vector1 : ", vector1)
print("Vector2 : ", vector2)
print("Addition: ", vector1 + vector2)
```

```python
#Vector multiplication
vector3 = np.array([[10,20,30], [5,6,10]])
print("Vector3: \n", vector3)

vector4 = np.array([[40,50], [20,25],[30,40]])
print("\nVector4: \n", vector4)

#multiplication using @
mul = vector3 @ vector4
print("\nMultiplication using @: \n", multi)
#multiplication using np.matmul
multi = np.matmul(vector3, vector4)
print("\nMultiplication using matmul: \n",multi)
```
