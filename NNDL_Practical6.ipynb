```python


!pip -q install kagglehub

import os
import glob
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
import kagglehub

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error

from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, LSTM, RepeatVector, TimeDistributed, Dense
from tensorflow.keras.callbacks import EarlyStopping

import warnings
warnings.filterwarnings("ignore")

print("TensorFlow Version:", tf.__version__)


print("\nDownloading Kaggle dataset...")

path = kagglehub.dataset_download(
    "aditya230688/jhts-forecasting"
)

print("Dataset downloaded successfully.")
print("Path:", path)



csv_files = glob.glob(
    os.path.join(path, "**", "*.csv"),
    recursive=True
)

if len(csv_files) == 0:
    raise FileNotFoundError(
        "No CSV file found in the downloaded Kaggle dataset."
    )

print("\nCSV Files Found:")

for i, file in enumerate(csv_files):
    print(i, ":", os.path.basename(file))


# Automatically select the CSV containing
# electricity_consumption

selected_file = None

for file in csv_files:

    temp_df = pd.read_csv(file, nrows=5)

    if "electricity_consumption" in temp_df.columns:
        selected_file = file
        break


if selected_file is None:

    raise ValueError(
        "Could not find a CSV containing "
        "'electricity_consumption' column."
    )


print("\nSelected File:")
print(selected_file)

df = pd.read_csv(selected_file)

print("\nDataset Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nFirst 5 Rows:")
print(df.head())


VALUE_COL = "electricity_consumption"

if VALUE_COL not in df.columns:

    raise ValueError(
        "Column 'electricity_consumption' not found."
    )


# Convert electricity consumption to numeric

df[VALUE_COL] = pd.to_numeric(
    df[VALUE_COL],
    errors="coerce"
)


# Remove invalid values

df = df.dropna(
    subset=[VALUE_COL]
)


# Reset index

df = df.reset_index(drop=True)


print("\nAfter preprocessing:", df.shape)

print("\nMissing Values:")

print(
    df[VALUE_COL].isnull().sum()
)


plt.figure(figsize=(15, 5))

plt.plot(
    df[VALUE_COL],
    label="Electricity Consumption"
)

plt.title(
    "Original Electricity Consumption"
)

plt.xlabel("Time / Record")

plt.ylabel("Electricity Consumption")

plt.grid(True)

plt.legend()

plt.show()



energy_values = df[
    [VALUE_COL]
].values


split_index = int(
    len(energy_values) * 0.80
)

train_values = energy_values[
    :split_index
]

test_values = energy_values[
    split_index:
]


print(
    "\nTraining Data Shape:",
    train_values.shape
)

print(
    "Testing Data Shape:",
    test_values.shape
)



scaler = MinMaxScaler(
    feature_range=(0, 1)
)


# Fit ONLY on training data

train_scaled = scaler.fit_transform(
    train_values
)

test_scaled = scaler.transform(
    test_values
)


print("\nOriginal Data:")

print(
    train_values[:5]
)


print("\nNormalized Data:")

print(
    train_scaled[:5]
)



SEQUENCE_LENGTH = 24


def create_sequences(
    data,
    sequence_length
):

    sequences = []

    for i in range(
        len(data) - sequence_length + 1
    ):

        sequences.append(
            data[
                i:i + sequence_length
            ]
        )

    return np.array(sequences)


X_train = create_sequences(
    train_scaled,
    SEQUENCE_LENGTH
)

X_test = create_sequences(
    test_scaled,
    SEQUENCE_LENGTH
)


print(
    "\nTraining Sequence Shape:",
    X_train.shape
)

print(
    "Testing Sequence Shape:",
    X_test.shape
)


timesteps = X_train.shape[1]

features = X_train.shape[2]


# Input

inputs = Input(
    shape=(timesteps, features)
)


# Encoder

encoded = LSTM(
    64,
    activation="tanh",
    return_sequences=False
)(inputs)


# Repeat encoded vector

repeated = RepeatVector(
    timesteps
)(encoded)


# Decoder

decoded = LSTM(
    64,
    activation="tanh",
    return_sequences=True
)(repeated)


# Output

outputs = TimeDistributed(
    Dense(features)
)(decoded)


# Create model

autoencoder = Model(
    inputs,
    outputs
)


print("\nLSTM AUTOENCODER MODEL:")

autoencoder.summary()


autoencoder.compile(
    optimizer="adam",
    loss="mse"
)


early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=5,
    restore_best_weights=True
)


history = autoencoder.fit(
    X_train,
    X_train,
    epochs=15,
    batch_size=32,
    validation_split=0.10,
    shuffle=False,
    callbacks=[early_stopping],
    verbose=1
)


plt.figure(figsize=(10, 5))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.title(
    "LSTM Autoencoder Training and Validation Loss"
)

plt.xlabel("Epoch")

plt.ylabel("MSE Loss")

plt.legend()

plt.grid(True)

plt.show()


X_test_pred = autoencoder.predict(
    X_test,
    verbose=0
)


print(
    "\nOriginal Test Shape:",
    X_test.shape
)

print(
    "Reconstructed Shape:",
    X_test_pred.shape
)



reconstruction_error = np.mean(
    np.square(
        X_test - X_test_pred
    ),
    axis=(1, 2)
)


print(
    "\nFirst 10 Reconstruction Errors:"
)

print(
    reconstruction_error[:10]
)


X_train_pred = autoencoder.predict(
    X_train,
    verbose=0
)


train_reconstruction_error = np.mean(
    np.square(
        X_train - X_train_pred
    ),
    axis=(1, 2)
)



threshold = (
    np.mean(
        train_reconstruction_error
    )
    +
    3 * np.std(
        train_reconstruction_error
    )
)


# Detect anomalies

anomalies = (
    reconstruction_error > threshold
)


print(
    "\nAnomaly Threshold:",
    threshold
)

print(
    "Number of Anomalies:",
    np.sum(anomalies)
)

plt.figure(figsize=(15, 6))

plt.plot(
    reconstruction_error,
    label="Reconstruction Error"
)

plt.axhline(
    threshold,
    linestyle="--",
    label="Anomaly Threshold"
)

plt.scatter(
    np.where(anomalies)[0],
    reconstruction_error[anomalies],
    s=40,
    label="Anomaly"
)

plt.title(
    "Energy Consumption Anomaly Detection"
)

plt.xlabel("Test Sequence")

plt.ylabel("Reconstruction Error")

plt.legend()

plt.grid(True)

plt.show()


sequence_index = 0


original_sequence = scaler.inverse_transform(
    X_test[sequence_index]
)


reconstructed_sequence = scaler.inverse_transform(
    X_test_pred[sequence_index]
)


plt.figure(figsize=(12, 5))

plt.plot(
    original_sequence.flatten(),
    label="Original",
    linewidth=2
)

plt.plot(
    reconstructed_sequence.flatten(),
    label="Reconstructed",
    linestyle="--",
    linewidth=2
)

plt.title(
    "Original vs Reconstructed Energy Consumption"
)

plt.xlabel("Time Step")

plt.ylabel("Energy Consumption")

plt.legend()

plt.grid(True)

plt.show()


mse = mean_squared_error(
    X_test.flatten(),
    X_test_pred.flatten()
)


mae = mean_absolute_error(
    X_test.flatten(),
    X_test_pred.flatten()
)


rmse = np.sqrt(mse)


max_error = np.max(
    np.abs(
        X_test - X_test_pred
    )
)


if max_error == 0:

    accuracy = 100

else:

    accuracy = (
        1 -
        (mae / max_error)
    ) * 100


accuracy = max(
    0,
    min(100, accuracy)
)


print("\n===================================")

print("       MODEL EVALUATION")

print("===================================")

print(
    "Mean Squared Error (MSE):",
    mse
)

print(
    "Mean Absolute Error (MAE):",
    mae
)

print(
    "Root Mean Squared Error (RMSE):",
    rmse
)

print(
    "Reconstruction Accuracy:",
    round(accuracy, 2),
    "%"
)


print("\n===================================")

print("          FINAL RESULT")

print("===================================")

print(
    "LSTM Autoencoder trained successfully."
)

print(
    "Dataset: JHTS Forecasting"
)

print(
    "Input Column:",
    VALUE_COL
)

print(
    "Input Sequence Length:",
    SEQUENCE_LENGTH
)

print(
    "Training Sequences:",
    len(X_train)
)

print(
    "Testing Sequences:",
    len(X_test)
)

print(
    "Anomalies Detected:",
    np.sum(anomalies)
)

print(
    "Accuracy:",
    round(accuracy, 2),
    "%"
)

print("===================================")
```
